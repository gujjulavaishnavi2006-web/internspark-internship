```python
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files
```

```python
uploaded = files.upload()
```

```python
file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

print(df.head())
```

```python
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
df.drop_duplicates(inplace=True)

print("Duplicates Removed")
```

```python
print(df.describe())
```

```python
df.head(100)[df.columns[1]].hist()

plt.title("Histogram")

plt.show()
```

```python
df[df.columns[1]].value_counts().head(5).plot(kind='bar')

plt.show()
```

```python
df[df.columns[1]].value_counts().head(5).plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.show()
```
