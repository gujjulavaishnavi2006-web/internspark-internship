```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'], dayfirst=True)
```

```python
total_revenue = df['Sales'].sum()

print("Total Revenue:", total_revenue)
```

```python
average_order_value = df['Sales'].mean()

print("Average Order Value:", average_order_value)
```

```python
monthly_sales = df.groupby(df['Order_Date'].dt.to_period('M'))['Sales'].sum()

monthly_sales.plot(figsize=(12,6))

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid(True)

plt.show()
```

```python
region_sales = df.groupby('Region')['Sales'].sum()

region_sales.plot(kind='bar', figsize=(8,5))

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")

plt.show()
```

```python
category_sales = df.groupby('Category')['Sales'].sum()

category_sales.plot(kind='bar', figsize=(8,5))

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")

plt.show()
```

```python
top_products = df.groupby('Product_Name')['Sales'].sum().sort_values(ascending=False).head(10)

top_products.plot(kind='barh', figsize=(10,6))

plt.title("Top 10 Products")
plt.xlabel("Sales")

plt.show()
```

```python
worst_products = df.groupby('Product_Name')['Sales'].sum().sort_values().head(10)

worst_products.plot(kind='barh', figsize=(10,6))

plt.title("Worst 10 Products")
plt.xlabel("Sales")

plt.show()
```

```python
subcategory_sales = df.groupby('Sub_Category')['Sales'].sum().sort_values()

subcategory_sales.plot(kind='barh', figsize=(10,8))

plt.title("Sales by Sub-Category")
plt.xlabel("Sales")

plt.show()
```
